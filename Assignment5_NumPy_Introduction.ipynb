## Solution 1 — Elementwise operations

## Solution 2 — Reverse an array and find modes with their indices

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed:", arr[::-1])

def modes_and_indices(array):
    values, counts = np.unique(array, return_counts=True)
    max_count = counts.max()
    modes = values[counts == max_count]
    return modes, {int(value): np.flatnonzero(array == value) for value in modes}

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1,1,1,2,3,4,2,4,3,3])
for name, array in [("x", x), ("y", y)]:
    mode_values, locations = modes_and_indices(array)
    print(name, "mode(s):", mode_values, "indices:", locations)
```

## Solution 4 — Evenly spaced values and transpose

```python
samarth = np.linspace(10, 100, 25)
print("Array:", samarth)
print("Dimensions:", samarth.ndim)
print("Shape:", samarth.shape)
print("Elements:", samarth.size)
print("Element dtype:", samarth.dtype)
print("Bytes:", samarth.nbytes)

column_view = samarth.reshape(25, 1).T
print("Transpose through reshape + transpose:\n", column_view)
print("T attribute on a 1-D array is unchanged:", samarth.T.shape)
print("To use T meaningfully, first make it 2-D:", samarth.reshape(1, -1).T.shape)
```

## Solution 5 — Statistics, reshape and resize

```python
ucs420_samarth = np.array([[10,20,30,40], [50,60,70,80], [90,15,20,35]])
print("Mean:", np.mean(ucs420_samarth))
print("Median:", np.median(ucs420_samarth))
print("Max:", np.max(ucs420_samarth))
print("Min:", np.min(ucs420_samarth))
print("Unique:", np.unique(ucs420_samarth))

reshaped_ucs420_samarth = ucs420_samarth.reshape(4, 3)
resized_ucs420_samarth = np.resize(ucs420_samarth, (2, 3))
print("Reshaped 4x3:\n", reshaped_ucs420_samarth)
print("Resized 2x3 (repeats values if needed):\n", resized_ucs420_samarth)
```

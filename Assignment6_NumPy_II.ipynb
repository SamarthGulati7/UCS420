## Solution 1 — Correct sensor bias and convert units

```python
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected_c = temperature + 2
fahrenheit = corrected_c * 9 / 5 + 32
above_32 = corrected_c[corrected_c > 32]
print("Corrected Celsius:", corrected_c)
print("Corrected Fahrenheit:", fahrenheit)
print("Corrected readings above 32 C:", above_32)
print("Count above 32 C:", np.count_nonzero(corrected_c > 32))
print("Vectorization applies operations to an array in compiled numerical routines, reducing Python-level loop overhead. Boolean indexing selects matching values directly and keeps the selection concise." )
```

## Solution 2 — Steps matrix and axis reductions

```python
steps = np.array([[5000,6200,7100], [8000,7500,9000], [4500,5100,4800], [9000,8500,9500]])
print("Total:", steps.sum())
print("Mean:", steps.mean())
print("Max:", steps.max())
print("Min:", steps.min())
print("Per day (axis=0):", steps.sum(axis=0))
print("Per user (axis=1):", steps.sum(axis=1))
max_position = np.unravel_index(steps.argmax(), steps.shape)
print("Max location (zero-based user, day):", max_position)
```

## Solution 3 — Slices, copies, reshape and flattening

```python
original = np.array([1,2,3,4,5,6])
subset = original[1:5]
subset[0] = 999
print("View changed original:", original)
print("subset:", subset)

copied = original[1:5].copy()
copied[0] = 500
print("Original after editing copy:", original)
print("Copied slice:", copied)

matrix = np.arange(1,13).reshape(3,4)
print("Matrix:\n", matrix)
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2, columns 2-3 (one-based):\n", matrix[0:2, 1:3])
flat_copy = matrix.flatten()
flat_view = matrix.ravel()
flat_view[0] = -1
flat_copy[1] = -2
print("After modifying ravel and flatten:\n", matrix)
print("flatten() returns an independent copy; ravel() usually returns a view when possible.")
print("shape, ndim, size, dtype:", matrix.shape, matrix.ndim, matrix.size, matrix.dtype)
```

## Solution 4 — Ordinary Least Squares

```python
X = np.array([[6,70,3], [5,50,6], [8,80,2], [4,30,8]], dtype=float)
y = np.array([40,65,30,85], dtype=float)
print("X shape/dimensions:", X.shape, X.ndim)
print("X transpose:\n", X.T)
print("X.T @ X:\n", X.T @ X)

# The normal equation is defined only if X.T @ X is invertible.
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print("Coefficients [sleep, activity, stress]:", beta)
new_user = np.array([5,40,7])
prediction = new_user @ beta
print("Predicted assistance score:", prediction)
print("Each coefficient is the fitted change in predicted score per one-unit increase in its feature, holding the other two fixed. With only four observations and three features (and no intercept), these estimates should be interpreted cautiously.")
```
